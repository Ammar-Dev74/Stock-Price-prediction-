# IMPORTING REQUIRED LIBRARIES

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.layers import Dense, Dropout, LSTM
from tensorflow.keras.models import Sequential
from tensorflow.keras.models import load_model

```

# LOAD THE CSV FILE INTO DATAFRAME 

```python
csv_file_path =("C:/Users/Ammar/Desktop/stock price prediction/data.csv")

# Load the CSV file into a DataFrame
data = pd.read_csv("C:/Users/Ammar/Desktop/stock price prediction/data.csv")
```

# DROP ANY ROWS WITH MISSING VALUES

```python
data.dropna(inplace=True)
```

# RESET THE INDEX OF THE DATAFRAME

```python

data.reset_index(drop=True, inplace=True)

# Print the cleaned data
print(data)
```

# SPLIT DATA FOR TRAIN AND TEST

```python
#Split 80% data for training and 20% data for testing
data_train = data['Close'][0:int(len(data) * 0.80)]
data_test = data['Close'][int(len(data) * 0.80):]

```

# TRAIN DATA

```python
data_train.shape[0]
```

# TEST DATA

```python
data_test.shape[0]
```

# SCALING THE TRAINING DATA

```python
scaler = MinMaxScaler(feature_range=(0,1))
data_train_scaled = scaler.fit_transform(data_train.values.reshape(-1,1))

data_train_scaled
```

# PREPARE DATA FOR LSTM

```python
x_train = []
y_train = []

for i in range(100,(data_train_scaled.shape[0])):
    x_train.append(data_train_scaled[i-100:i])
    y_train.append(data_train_scaled[i,0])



      
```

# RESHAPE INPUT DATA FOR LSTM

```python
x_train, y_train, = np.array(x_train), np.array(y_train)  
x_train = np.reshape(x_train, (x_train.shape[0], x_train.shape[1], 1))
```

# BUILD AND TRAIN THE LSTM MODEL

```python
lstm_model = Sequential()
lstm_model.add(LSTM(units = 50, activation = 'relu',return_sequences =True,
               input_shape = ((x_train.shape[1],1))))
lstm_model.add(Dropout(0.2))  

lstm_model.add(LSTM(units = 50, activation = 'relu' , return_sequences =False,
                   ))
               
lstm_model.add(Dropout(0.2))  


lstm_model.add(Dense(units =1))

lstm_model.compile(optimizer = 'adam', loss = 'mean_squared_error')
               

```

# COMPILE AND TRAIN THE MODEL

```python
lstm_model.fit(x_train,y_train, epochs = 20,batch_size =32, verbose =1)
```

# MODEL SUMMARY

```python
lstm_model.summary()
```

# PREPARE TEST DATA FOR PREDICTION

```python
#Take the last 100 days of training data & concatenate with test data.
pas_100_days = data_train.tail(100)
```

# SCALING TEST DATA

```python
data_test_combined= pd.concat([pas_100_days,data_test], ignore_index=True)
data_test_scaled = scaler.fit_transform(data_test_combined.values.reshape(-1,1))
```

# PREPARE TEST DATA FOR LSTM PREDICTION

```python
x_test = []
y_test = []
for i in range(100,(data_test_scaled.shape[0])):
    x_test.append(data_test_scaled[i-100:i , 0])
    y_test.append(data_test_scaled[i,0])

x_test = np.array(x_test)
x_test = np.reshape(x_test, (x_test.shape[0], x_test.shape[1], 1))  # Reshape for LSTM input
```

# MAKE PREDICTIONS

```python
lstm_predictions = lstm_model.predict(x_test)
lstm_predictions = scaler.inverse_transform(lstm_predictions)
lstm_predictions = lstm_predictions.reshape(-1, 1)
print(lstm_predictions) 

```

# CALCULATE MAE, MSE, RMSE, R²:

```python
def print_metrics(y_true, y_pred, model_name):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    
    print(f"Metrics for {model_name}:")
    print(f"Mean Absolute Error (MAE): {mae:.4f}")
    print(f"Mean Squared Error (MSE): {mse:.4f}")
    print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
    print(f"R-squared (R²): {r2:.4f}")
    print("-" * 50)
```
